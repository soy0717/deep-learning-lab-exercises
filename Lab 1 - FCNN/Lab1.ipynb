# Implementation of a fully connected neural network to add 4-digit numbers from scratch

```python
import numpy as np
```

```python
# activation function
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x > 0).astype(float)

# Mean Squared Error
def mse(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

def mse_derivative(y_true, y_pred):
    return 2 * (y_pred - y_true) / y_true.shape[0]
```

```python
# generate the dataset
num_samples = 1000
x_train = np.random.rand(num_samples, 2)
y_train = (x_train[:, 0] + x_train[:, 1]).reshape(-1, 1)
```

```python
input_size = 2
hidden_size = 8
output_size = 1

np.random.seed(42)
```

```python
# weights and biases
w1 = np.random.randn(input_size, hidden_size) * 0.1
b1 = np.zeros((1, hidden_size))

w2 = np.random.randn(hidden_size, output_size) * 0.1
b2 = np.zeros((1, output_size))
```

```python
epochs = 100
batch_size = 32
lr = 0.01
```

```python
for epoch in range(epochs):
    perm = np.random.permutation(num_samples)
    x_train = x_train[perm]
    y_train = y_train[perm]

    for i in range(0, num_samples, batch_size):
        X_batch = x_train[i:i+batch_size]
        y_batch = y_train[i:i+batch_size]

        # Forward pass
        z1 = np.dot(X_batch, w1) + b1
        a1 = relu(z1)
        z2 = np.dot(a1, w2) + b2
        y_pred = z2

        # Loss
        loss = mse(y_batch, y_pred)

        # Backpropagation
        dloss = mse_derivative(y_batch, y_pred)     # (batch_size, 1)
        dw2 = np.dot(a1.T, dloss)                   # (hidden, 1)
        db2 = np.sum(dloss, axis=0, keepdims=True)  # (1, 1)

        da1 = np.dot(dloss, w2.T)                   # (batch_size, hidden)
        dz1 = da1 * relu_derivative(z1)             # (batch_size, hidden)

        dw1 = np.dot(X_batch.T, dz1)                # (input, hidden)
        db1 = np.sum(dz1, axis=0, keepdims=True)    # (1, hidden)

        # Update weights
        w2 -= lr * dw2
        b2 -= lr * db2
        w1 -= lr * dw1
        b1 -= lr * db1

    if epoch % 10 == 0 or epoch == epochs - 1:
        print(f"Epoch {epoch}, Loss: {loss:.4f}")
```

```python
def predict(x):
    z1 = np.dot(x, w1) + b1
    a1 = relu(z1)
    z2 = np.dot(a1, w2) + b2
    return z2
```

```python
test_input1 = np.array([[1, 2], [0.3, 0.4]])
pred_sum1 = predict(test_input1)
```

```python
print("Predicted sums:")
print(pred_sum1)
print("Actual sums:")
print(np.sum(test_input1, axis=1, keepdims=True))
```
